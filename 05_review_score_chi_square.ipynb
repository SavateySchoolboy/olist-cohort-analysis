```python
# χ²-тест независимости: сравниваем группы оценок 1–3 и 4–5.
# В строках таблицы — не вернулись и вернулись; в столбцах — группы оценок.
# Числа получены запросом 05_hypotheses_reviews.sql. correction=False — тест без поправки Йейтса.
from scipy.stats import chi2_contingency
no_returned_1_3 = 10189 - 177
no_returned_4_5 = 38417 - 793
chi2, p, dof, expected = chi2_contingency([[no_returned_1_3, no_returned_4_5], [177, 793]], correction=False)
print(chi2, p)
```

```python
# n0 и x0 — размер группы 1–3 и число вернувшихся; n1 и x1 — для группы 4–5.
# p0 и p1 — доли возврата. Разница diff = p0 − p1 указана в долях.
n0, x0 = 10189, 177    
n1, x1 = 38417, 793

p0 = x0 / n0
p1 = x1 / n1
diff = p0 - p1
```

```python
# Стандартная ошибка разницы долей по наблюдаемым долям обеих групп.
# Округление ниже меняет только отображение, само значение se сохраняет точность.
se = ((p0 * (1 - p0)/n0) + (p1 * (1 - p1)/n1))**0.5
round(se, 3)
```

```python
# Приближённый 95%-й интервал разницы p0 − p1 по нормальному приближению.
# Границы даны в долях; умножение на 100 переводит их в процентные пункты.
low = diff - 1.96 * se
high = diff + 1.96 * se
di = [low, high]
di
```

```python
# Общая доля вернувшихся для приближённого расчёта MDE.
# Здесь p получает новое значение: это общая доля, а не p-value из χ²-теста.
# 1.96 соответствует двустороннему уровню значимости 5%, 0.84 — мощности около 80%.
p = (x0 + x1) / (n1 + n0)
z_sum = 1.96 + 0.84
z_a_2 = 1.96
z_b = 0.84
```

```python
# Приближённый MDE в долях при текущих размерах групп и общей доле возврата.
# Умножение на 100 переводит результат в процентные пункты.
mde = z_sum * ((p * (1 - p)) * ((1/n0) + (1/n1)))**0.5
mde
```
